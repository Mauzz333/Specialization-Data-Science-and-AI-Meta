Realice las operaciones dadas para una regresión lineal simple y visualizar los datos dados con **xs** y **ys**. Muestre el gráfico para analizar el comportamiento de los datos.

$Recordando$ $la$ $Regresión$ $Lineal$ $Simple:$

$
\overline{y} = β\overline{x} + α
$

Entrene una red neuronal para hacer coincidir **xs** con **ys**.

De modo que un predictor para un nuevo valor de X proporcione un valor flotante muy cercano a la respuesta deseada, es decir, **print(modelo.predict([X]))** debe dar un resultado satisfactorio.

La infraestructura de prueba, espera un modelo entrenado que acepte una forma de entrada de [1].

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Que datos tengo? check
# Que algoritmo / modelo voy a utilizar
# Como voy a medir el error
```

```python
# Que datos tengo?
# xs = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)

xs = np.arange(-1, 5, dtype=float)
print(xs)
ys= np.array([-4.0, 1.0, 6.0, 11.0, 16.0, 21.0], dtype=float)
print(ys)
```

```python
plt.plot(xs, ys, 'ro')
plt.show()
```

```python
# Que algoritmo / modelo voy a utilizar

regresion_lineal = np.polyfit(xs,ys, 1)

beta, alpha = regresion_lineal

print(f'La ecuacion de la recta es: y = {regresion_lineal[0]}x + {regresion_lineal[1]}')
```

```python
plt.plot(xs, ys, 'o', color = 'green', label="datos")
plt.plot(xs, beta * xs + alpha, color = 'blue', label = 'Ajuste a la regresion Lineal')
plt.legend()
plt.axhline(y=0, color='orange')
plt.axvline(x=0, color='orange')
plt.xlabel('Tamaño de la casa')
plt.ylabel('Precio de la casa')
plt.title('Regresion Lineal')
plt.grid(True)
plt.show()
```

```python
capa = tf.keras.layers.Dense(input_shape=[1], units=1)
modelo = tf.keras.Sequential([capa])
modelo.compile(
    optimizer='sgd',
    loss='mean_squared_error'
)
```

```python
callback = tf.keras.callbacks.EarlyStopping(monitor='loss', patience=10, min_delta=0.1)
hist = modelo.fit(xs, ys, epochs=500, verbose=True, callbacks=[callback])
modelo, hist
```

```python
print('hola, voa comenzaar a entrenar')
hist = modelo.fit(xs, ys, epochs=500, verbose=True)
modelo, hist
```

```python
valor_a_predecir = float(input('Ingresa cuantas habitaciones tiene la casa para conocer su precio: '))

prediccion = modelo.predict(np.array([[valor_a_predecir]]))


print(f'El precio de la casa con {valor_a_predecir} habitaciones es: {prediccion}')
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
Celsius = np.array([-40.0, -10.0, 0.0, 8.0, 15.0, 22.0, 38.0], dtype=float)
print(Celsius)
Farahnhait= np.array([-40.0, 14.0, 32.0, 46.0, 59.0, 72.0, 100.00], dtype=float)
print(Farahnhait)
```

```python
plt.plot(Celsius, Farahnhait, 'ro')
plt.show()
```

```python
# Que algoritmo / modelo voy a utilizar

regresion_lineal = np.polyfit(Celsius,Farahnhait, 1)

beta, alpha = regresion_lineal

print(f'La ecuacion de la recta es: y = {regresion_lineal[0]}x + {regresion_lineal[1]}')
```

```python
plt.plot(Celsius, Farahnhait, 'o', color = 'green', label="datos")
plt.plot(Celsius, beta * Celsius + alpha, color = 'blue', label = 'Ajuste a la regresion Lineal')
plt.legend()
plt.axhline(y=0, color='orange')
plt.axvline(x=0, color='orange')
plt.xlabel('Celsius')
plt.ylabel('Fahrenheit')
plt.title('Regresion Lineal')
plt.grid(True)
plt.show()
```

```python
capa = tf.keras.layers.Dense(input_shape=[1], units=1)
modelo = tf.keras.Sequential([capa])
modelo.compile(
    optimizer='sgd',
    loss='mean_absolute_error'
)
```

```python
callback = tf.keras.callbacks.EarlyStopping(monitor='loss', patience=20, min_delta=0.001)
hist = modelo.fit(Celsius, Farahnhait, epochs=4200, verbose=True, callbacks=[callback])
modelo, hist
```

```python
print('hola, vamos a comenzaar a entrenar')
hist = modelo.fit(Celsius, Farahnhait, epochs=500, verbose=True)
modelo, hist
```

```python
valor_a_predecir = float(input('Ingresa la temperatura en grados Celsius: : '))

prediccion = modelo.predict(np.array([[valor_a_predecir]]))


print(f'La temperatura es de {valor_a_predecir}°C equivale a aproximadamente: {prediccion}')
```
