## 2. Limpieza de datos

```python
# 1. Eliminar filas donde los nulos son insignificantes
df_netflix = df_netflix.dropna(subset=['date_added', 'rating', 'duration'])

# 2. Imputar nulos en las columnas con faltantes masivos
df_netflix['director'] = df_netflix['director'].fillna('Unknown')
df_netflix['cast'] = df_netflix['cast'].fillna('No Cast Listed')
df_netflix['country'] = df_netflix['country'].fillna('Unknown Country')

# 3. Transformar formatos (Fechas)
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'].str.strip())

# 4. Verificar que todo quedó en 0
print(df_netflix.isnull().sum())
```

```python
df_netflix.head(5)
```

```python
df_netflix.dtypes
```

```python
#cambiar tipo de dato

df_netflix['release_year'] = df_netflix['release_year'].astype(int)
df_netflix['type'] = df_netflix['type'].astype('category')
df_netflix['country'] = df_netflix['country'].astype('category')
df_netflix['rating'] = df_netflix['rating'].astype('category')
```

```python
df_netflix.dtypes
```

```python
# 1. Crear la columna para los minutos de las películas
# Extraemos el número solo cuando el tipo es 'Movie'
df_netflix['duration_min'] = df_netflix.loc[df_netflix['type'] == 'Movie', 'duration'].str.replace(' min', '', case=False)
df_netflix['duration_min'] = df_netflix['duration_min'].astype(float) # Usamos float por si quedan nulos

# 2. Crear la columna para las temporadas de las series
# Limpiamos las palabras 'Season' o 'Seasons' usando una expresión regular
df_netflix['seasons'] = df_netflix.loc[df_netflix['type'] == 'TV Show', 'duration'].str.replace(r'\s*Seasons?$', '', case=False, regex=True)
df_netflix['seasons'] = df_netflix['seasons'].astype(float)

# 3. Verificamos cómo quedaron nuestras nuevas columnas numéricas
print(df_netflix[['type', 'duration', 'duration_min', 'seasons']].head())
```

```python
df_netflix.head(10)
```

```python
df_netflix['duration_min'] = df_netflix['duration_min'].fillna(0)
df_netflix['seasons'] = df_netflix['seasons'].fillna(0)
```

## 3. Exploración de datos

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
columnas_numericas = ['date_added', 'release_year', 'duration_min', 'seasons']
df_netflix[columnas_numericas].describe()
```

```python
plt.figure(figsize=(15, 9))

for i, col in enumerate(columnas_numericas, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df_netflix[col], kde=True, bins=30, color='teal')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
    plt.title(f'Distribución de {col}')
    plt.grid(axis= 'y', alpha = 0.75 )

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 9))

for i, col in enumerate(columnas_numericas, 1): #i = index se refiere
  plt.subplot(2, 3, i)
  sns.boxplot(x=df_netflix[col], color = 'yellow')
  plt.xlabel(col)
  plt.ylabel('Frecuencia')
  plt.title(f'Diagrama de caja y bigotes {col}')
  plt.grid(axis= 'x', alpha = 0.75 )

plt.tight_layout()
plt.show()
```

```python
# Ver cantidad de registros iniciales
print(f'Cantidad de datos iniciales: {len(df_netflix)}')
```

```python
# Calculo del IQR
df_copy = df_netflix.copy()

for col in columnas_numericas:
  Q1 = df_copy[col].quantile(0.25)
  Q3 = df_copy[col].quantile(0.75)
  IQR = Q3 - Q1

  lim_inf_iqr = Q1 - 1.5 * IQR
  lim_sup_iqr = Q3 + 1.5 * IQR

  cant_inicial = len(df_copy)
  df_copy = df_copy[(df_copy[col] >= lim_inf_iqr) & (df_copy[col] <= lim_sup_iqr)]
  cant_final = len(df_copy)
  print(f'Cantidad de datos con outliers eliminados en {col}: {cant_inicial - cant_final}')
```

```python
# 1. Aseguramos tener el año de adición como número entero
df_netflix['year_added'] = df_netflix['date_added'].dt.year

# 2. Creamos una lista con las columnas estrictamente numéricas que nos interesan
columnas_num_second = ['year_added', 'release_year', 'duration_min', 'seasons']

# 3. Calculamos la correlación filtrando SOLO esas columnas
matriz_corr = df_netflix[columnas_num_second].corr(numeric_only=True)


plt.figure(figsize=(8, 6))
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f")
plt.title('Matriz de Correlación de Variables Numéricas')
plt.show()
```

```python
df_copy = df_netflix.copy()

# 2. Creamos la columna 'year_added' extrayendo el año de la fecha limpiada
df_copy['year_added'] = df_copy['date_added'].dt.year

# --- AQUÍ EMPIEZA TU CÓDIGO ORIGINAL ---
metricas = ['year_added', 'release_year', 'duration_min', 'seasons']

# Ojo aquí: Tu lista tiene 6 comparaciones. 
# Si pones 2 filas y 5 columnas (nrows=2, ncols=5), tendrás 10 subplots en total. 
# Está perfecto, pero te van a quedar 4 cuadros vacíos al final.
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(36, 12))
axes = axes.flatten()

# Pares de comparación por
comparaciones = [
    ('year_added', 'release_year'), ('year_added', 'duration_min'), ('year_added', 'seasons'),
    ('release_year', 'duration_min'), ('release_year', 'seasons'), ('duration_min', 'seasons')
]

for i, (col_x, col_y) in enumerate(comparaciones):
    # Usamos .dropna para que regplot no tenga problemas con los NaN lógicos
    data_filtrada = df_copy.dropna(subset=[col_x, col_y])
    
    sns.regplot(ax=axes[i], data=data_filtrada, x=col_x, y=col_y,
                scatter_kws={'alpha':0.3, 'color':'darkorchid'},
                line_kws={'color':'blue'})
    axes[i].set_title(f'{col_x.title()} vs {col_y.title()}')
    axes[i].grid(True, alpha=0.2)

# Opcional: Ocultamos los 4 subplots vacíos que te sobran para que se vea limpio
for j in range(len(comparaciones), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## 4.Visualización de datos

```python

plt.figure(figsize=(10, 5))
sns.set_style("whitegrid")

orden_rating = df_netflix['rating'].value_counts().index

ax = sns.countplot(
    data=df_netflix,
    x='rating',
    order=orden_rating,
    palette='mako', # Mantenemos tu paleta favorita
    hue='rating',
    legend=False
)


for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f'{int(height)}',
            (p.get_x() + p.get_width() / 2., height),
            ha='center', va='bottom',
            fontsize=10, fontweight='bold',
            xytext=(0, 5), textcoords='offset points'
        )

plt.title('Cantidad de Títulos en Netflix por Clasificación (Rating)', loc='left', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Clasificación de Edad')
plt.ylabel('Cantidad de Contenido')
sns.despine()
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(24, 6))

ax = plt.subplot(1, 2, 1)
orden = df_copy['type'].value_counts().index
sns.countplot(data=df_copy, y='type', order=orden, palette='viridis', hue='type', legend=False)
for p in ax.patches:
    width = p.get_width() # Obtenemos la cantidad exacta de la barra
    if width > 0:         # Nos aseguramos de que la barra no esté vacía
        ax.annotate(
            f'{int(width):,}',                 
            (width, p.get_y() + p.get_height() / 2), 
            ha='left',                          
            va='center',                        
            fontsize=11,                       
            fontweight='bold',                  
            xytext=(8, 0),                     
            textcoords='offset points'          
        )
plt.title('Distribución de Tipos de Contenido', fontsize=14)
plt.xlabel('Cantidad de propiedades')
plt.ylabel('Tipo de Contenido')
plt.grid(axis='x', linestyle='--', alpha=0.6)

plt.subplot(1, 2, 2)
sns.boxplot(data=df_copy, x='type', y='release_year', palette='viridis', hue='type', legend=False, showfliers=False)
plt.title('Año de Estreno por Tipo de Contenido')
plt.xlabel('Tipo de Contenido')
plt.ylabel('Año de Estreno')

plt.tight_layout()
plt.show()
```

```python
categoria = lambda x: 'Película' if x == 'Movie' else 'Serie'

df_copy['Tipo de Contenido'] = df_copy['type'].apply(categoria)
pie_data = df_copy['Tipo de Contenido'].value_counts()
pie_data

# Realizar gráfico de dona


colors_pie = ["#009B2290", "#117D8380"] # Azul = "Con Categoría", Gris = "Sin Categoría"
explode_val = 0.05 # Valor de separación

# Gráfico de pastel

plt.figure(figsize=(6, 6))
wedges, texts, autotexts = plt.pie(pie_data,
                                   labels=pie_data.index,
                                   autopct='%1.1f%%',
                                   startangle=90,
                                   colors=colors_pie,
                                   textprops={'fontsize': 12, 'fontweight': 'bold'},
                                   pctdistance=0.85, # Mueve el porcentaje hacia afuera
                                   explode=(explode_val, 0)) # Resalta ligeramente la porción principal

# Círculo blanco central (estilo moderno)

theta1, theta2 = wedges[0].theta1, wedges[0].theta2
center_angle = np.deg2rad((theta1 + theta2) / 2) # Obtención el ángulo medio

cx = explode_val * np.cos(center_angle) # Se Calcula nueva posición con trigonometría
cy = explode_val * np.sin(center_angle)

centre_circle = plt.Circle((cx, cy), 0.70, fc='white') # Se dibuja el circulo
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Proporción de Películas vs Series en Netflix', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.set_style("whitegrid")

# 1. Obtenemos los 10 países con más contenido para evitar saturar el gráfico
top_paises = df_copy['country'].value_counts().head(10).index
df_top_paises = df_copy[df_copy['country'].isin(top_paises)]

# 2. Graficamos agrupando por 'type' usando el parámetro hue
ax = sns.countplot(
    data=df_top_paises,
    y='country',
    hue='type',
    order=top_paises,
    palette='viridis' # Mantiene la consistencia de tus colores
)

# 3. El ciclo FOR para poner las cantidades exactas al lado de cada barra
for p in ax.patches:
    width = p.get_width()
    if width > 0:
        ax.annotate(
            f'{int(width)}',
            (width, p.get_y() + p.get_height() / 2),
            ha='left', va='center',
            fontsize=10, fontweight='bold',
            xytext=(5, 0), textcoords='offset points'
        )

plt.title('Producción de Películas vs Series de TV (Top 10 Países)', loc='left', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Cantidad de Títulos Registrados')
plt.ylabel('País')
plt.legend(title='Tipo de Contenido', loc='lower right')

# Le damos un 15% de espacio extra a la derecha para que no se corten los números
plt.xlim(0, df_top_paises['country'].value_counts().max() * 1.15)

sns.despine()
plt.tight_layout()
plt.show()
```
